### Enunciado
O objetivo desta tarefa é comparar as sequências de DNA de enzimas topoisomerase de três organismos diferentes: rato, hamster chinês e cavalo. A comparação será feita em dois níveis principais:
1. Verificação de proximidade entre dois organismos para três pares específicos de organismos: hamster-horse, hamster-rat, horse-rat.
2. Cálculo de distâncias entre os organismos usando várias métricas, como distância Manhattan, euclidiana, supremum e a similaridade de cosseno.

Os arquivos de entrada ("hamster.fasta", "horse.fasta" e "rat.fasta") contêm as sequências de DNA dessas enzimas.

```python
import string

def read_fasta(arq):
    seq = ''
    with open(arq) as f:
        f.readline()  # ignora a primeira linha do arquivo (cabeçalho)
        for line in f:
            seq += line.strip()  # concatenando as sequências
    return seq

# Leitura dos arquivos FASTA com as sequências de DNA dos organismos
fastas = []
fastas.append(read_fasta('hamster.fasta'))
fastas.append(read_fasta('horse.fasta'))
fastas.append(read_fasta('rat.fasta'))
organismos = ['hamster', 'horse', 'rat']

# Exemplo de leitura e armazenamento das sequências de DNA
```

### Exercício 1: Verificação Simples
A verificação simples consiste em comparar diretamente os caracteres das sequências de dois organismos, retornando a proporção de caracteres iguais nas duas sequências.

A função abaixo realiza essa comparação para duas sequências.

```python
def versimp(seq1, seq2):
    iguais = 0
    for i in range(len(seq1)):
        if(seq1[i] == seq2[i]):
            iguais += 1
    res = iguais / len(seq1)
    return "{:.4f}".format(res)

# Teste da função versimp entre duas sequências de exemplo
result_versimp = versimp(fastas[0], fastas[1])  # comparando hamster e horse
result_versimp

# A função retorna a proporção de caracteres iguais nas duas sequências.
```

### Exercício 2: Contagem de Ocorrências de Aminoácidos
Aqui, vamos contar a ocorrência de cada aminoácido (representado por letras de A a Z) nas sequências de DNA dos organismos.
Depois, vamos calcular as distâncias entre as sequências de aminoácidos usando diferentes métricas.

```python
def contaraa(seq):
    cont = []
    for x in string.ascii_uppercase:  # A-Z
        cont.append(seq.count(x))  # contagem de cada letra na sequência
    return cont

# Contagem das ocorrências de aminoácidos nas sequências
amino_counts = []
for seq in fastas:
    amino_counts.append(contaraa(seq))
amino_counts

# A lista de contagem mostra quantas vezes cada aminoácido (de A a Z) aparece nas sequências.
```

### Distâncias: Manhattan, Euclidiana e Supremum
Agora vamos calcular as distâncias entre as sequências usando as seguintes métricas:
1. Distância Manhattan: soma das diferenças absolutas entre as ocorrências de aminoácidos.
2. Distância Euclidiana: raiz quadrada da soma dos quadrados das diferenças das ocorrências.
3. Distância Supremum: a maior diferença absoluta entre as ocorrências.

Cada uma dessas distâncias será calculada entre os pares de sequências.

```python
def dmanhattan(seq1, seq2):
    soma = 0
    for i in range(len(seq1)):
        soma += abs(seq1[i] - seq2[i])  # soma das diferenças absolutas
    return "{:.4f}".format(soma)

def deuclidiana(seq1, seq2):
    soma = 0
    for i in range(len(seq1)):
        soma += (seq1[i] - seq2[i]) ** 2  # soma dos quadrados das diferenças
    return "{:.4f}".format(soma ** 0.5)  # raiz quadrada

def dsupremum(seq1, seq2):
    lista = [abs(seq1[i] - seq2[i]) for i in range(len(seq1))]  # calcula as diferenças
    return "{:.4f}".format(max(lista))  # maior diferença

# Exemplo de cálculo das distâncias entre as sequências hamster e horse
dist_manhattan = dmanhattan(amino_counts[0], amino_counts[1])
dist_euclidiana = deuclidiana(amino_counts[0], amino_counts[1])
dist_supremum = dsupremum(amino_counts[0], amino_counts[1])

dist_manhattan, dist_euclidiana, dist_supremum
```

### Similaridade de Cosseno
A similaridade de cosseno mede o quão semelhantes são duas sequências em termos de seu ângulo. Quanto mais próximas as sequências, maior será a similaridade.

Agora vamos calcular a similaridade de cosseno entre as sequências de DNA dos diferentes pares de organismos.

```python
def simcos(x, y):
    xy = sum(x[i] * y[i] for i in range(len(x)))  # soma do produto dos elementos
    x2 = sum(x[i] ** 2 for i in range(len(x)))  # soma dos quadrados
    y2 = sum(y[i] ** 2 for i in range(len(y)))  # soma dos quadrados
    return "{:.4f}".format(xy / ((x2 ** 0.5) * (y2 ** 0.5)))  # similaridade de cosseno

# Cálculo da similaridade de cosseno entre as sequências hamster e horse
sim_cosseno = simcos(amino_counts[0], amino_counts[1])
sim_cosseno
```

### Geração do Arquivo de Saída
Agora, vamos gerar um arquivo de saída contendo as comparações entre os pares de organismos, com as distâncias e similaridade de cosseno calculadas.
O arquivo será salvo em formato `.txt`.

```python
#Comparar todos os organismos
def comparar(fastas,opcao):
  res=''
  for m in range(len(fastas)):
    for n in range(len(fastas)):
      if(m<n):
        if(opcao=='versimp'):
          res=res+organismos[m]+','+organismos[n]+': '+versimp(fastas[m],fastas[n])+'\n'
        elif(opcao=='dmanhattan'):
          res=res+organismos[m]+','+organismos[n]+': '+str(dmanhattan(fastas[m],fastas[n]))+'\n'
        elif(opcao=='deuclidiana'):
          res=res+organismos[m]+','+organismos[n]+': '+str(deuclidiana(fastas[m],fastas[n]))+'\n'
        elif(opcao=='dsupremum'):
          res=res+organismos[m]+','+organismos[n]+': '+str(dsupremum(fastas[m],fastas[n]))+'\n'
        elif(opcao=='simcos'):
          res=res+organismos[m]+','+organismos[n]+': '+str(simcos(fastas[m],fastas[n]))+'\n'
  return res
```

```python
with open('saida01.txt','w') as arquivo:
  arquivo.write('Tarefa 1 - Similaridade de enzimas entre organismos\n')
  arquivo.write('Ana Luiza Moraes Octaviano\t11779931')
  arquivo.write('\nGiuliana Lopes Pola\t11779802')
  arquivo.write('\n\nExercício 01: verificação simples\n')
  arquivo.write(comparar(fastas,'versimp'))
  arquivo.write('\nExercício 2:\n')
  arquivo.write('\na.Contagem de ocorrências de cada aminoácido\n')
  arquivo.write('{:10}'.format(' ')+' '+' '.join('{:3}'.format(s) for s in string.ascii_uppercase)+'\n')

  naa=[]
  #naa=fastas
  for i in range(len(fastas)):
    arquivo.write('{:10}'.format(organismos[i])+' ')
    res=contaraa(fastas[i])
    arquivo.write(' '.join('{:3}'.format(str(int)) for int in res))
    arquivo.write('\n')
    naa.append(res)
  
  arquivo.write('\nb.Distância manhattan entre dois organismos\n')
  arquivo.write(comparar(naa,'dmanhattan'))

  arquivo.write('\n\nc.Distância euclidiana entre dois organismos\n')
  arquivo.write(comparar(naa,'deuclidiana'))

  arquivo.write('\n\nd.Distância supremum entre dois organismos\n')
  arquivo.write(comparar(naa,'dsupremum'))

  arquivo.write('\n\nd.Similiaridade de cosseno entre dois organismos\n')
  arquivo.write(comparar(naa,'simcos'))



  
```
