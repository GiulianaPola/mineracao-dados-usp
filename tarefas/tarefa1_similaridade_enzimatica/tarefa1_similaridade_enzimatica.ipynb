# 📘 Tarefa 1 - Similaridade de Enzimas entre Organismos  


## 🔎 Enunciado

Os três arquivos `hamster.fasta`, `horse.fasta` e `rat.fasta` contêm sequências de DNA da enzima **topoisomerase** de três organismos distintos: hamster chinês, cavalo e rato.

Seu objetivo é:

1. Comparar a similaridade das sequências entre todos os pares possíveis de organismos usando **verificação simples** (posição a posição).
2. Contar a frequência de ocorrência de cada aminoácido nas sequências e calcular diferentes **métricas de distância** entre os vetores gerados:
   - Distância **Manhattan**
   - Distância **Euclidiana**
   - Distância **Supremum**
   - **Similaridade de Cosseno**


## 📦 1. Leitura dos Arquivos FASTA

```python
import string

def read_fasta(arq):
    seq = ''
    with open(arq) as f:
        f.readline()
        for line in f:
            seq += line.strip()
    return seq
```

## 📄 2. Leitura das Sequências de DNA

```python
fastas = []
fastas.append(read_fasta('hamster.fasta'))
fastas.append(read_fasta('horse.fasta'))
fastas.append(read_fasta('rat.fasta'))
organismos = ['hamster', 'horse', 'rat']
```

## ✅ 3. Verificação Simples (Comparação Direta)

```python
def versimp(seq1, seq2):
    iguais = 0
    for i in range(len(seq1)):
        if seq1[i] == seq2[i]:
            iguais += 1
    res = iguais / len(seq1)
    return "{:.4f}".format(res)
```

## 🔢 4. Contagem de Ocorrências de Aminoácidos

```python
def contaraa(seq):
    cont = []
    for x in string.ascii_uppercase:
        cont.append(seq.count(x))
    return cont
```

## 📏 5. Distâncias Numéricas

### 🔸 a) Distância Manhattan

```python
def dmanhattan(seq1, seq2):
    soma = 0 
    for i in range(len(seq1)):
        soma += abs(seq1[i] - seq2[i])
    return "{:.4f}".format(soma)
```

### 🔸 b) Distância Euclidiana

```python
def deuclidiana(seq1, seq2):
    soma = 0 
    for i in range(len(seq1)):
        soma += (seq1[i] - seq2[i]) ** 2
    res = soma ** 0.5
    return "{:.4f}".format(res)
```

### 🔸 c) Distância Supremum

```python
def dsupremum(seq1, seq2):
    lista = []
    for i in range(len(seq1)):
        lista.append(abs(seq1[i] - seq2[i]))
    res = max(lista)
    return "{:.4f}".format(res)
```

## 🧮 6. Similaridade de Cosseno

```python
def simcos(x, y):
    xy = 0
    x2 = 0
    y2 = 0
    for i in range(len(x)):
        xy += x[i] * y[i]
        x2 += x[i] * x[i]
        y2 += y[i] * y[i]
    x2 = x2 ** 0.5
    y2 = y2 ** 0.5
    res = xy / (x2 * y2)
    return "{:.4f}".format(res)
```

## 🔁 7. Função Geral de Comparação

```python
def comparar(fastas, opcao):
    res = ''
    for m in range(len(fastas)):
        for n in range(len(fastas)):
            if m < n:
                if opcao == 'versimp':
                    res += f"{organismos[m]},{organismos[n]}: {versimp(fastas[m], fastas[n])}\n"
                elif opcao == 'dmanhattan':
                    res += f"{organismos[m]},{organismos[n]}: {dmanhattan(fastas[m], fastas[n])}\n"
                elif opcao == 'deuclidiana':
                    res += f"{organismos[m]},{organismos[n]}: {deuclidiana(fastas[m], fastas[n])}\n"
                elif opcao == 'dsupremum':
                    res += f"{organismos[m]},{organismos[n]}: {dsupremum(fastas[m], fastas[n])}\n"
                elif opcao == 'simcos':
                    res += f"{organismos[m]},{organismos[n]}: {simcos(fastas[m], fastas[n])}\n"
    return res
```

## 📤 8. Escrita dos Resultados em Arquivo

```python
with open('saida.txt','w') as arquivo:
    arquivo.write('Tarefa 1 - Similaridade de enzimas entre organismos\n')
    arquivo.write('Ana Luiza Moraes Octaviano\t11779931')
    arquivo.write('\nGiuliana Lopes Pola\t11779802')
    arquivo.write('\n\nExercício 01: verificação simples\n')
    arquivo.write(comparar(fastas, 'versimp'))

    arquivo.write('\nExercício 2:\n')
    arquivo.write('\na.Contagem de ocorrências de cada aminoácido\n')
    arquivo.write('{:10}'.format(' ') + ' ' + ' '.join('{:3}'.format(s) for s in string.ascii_uppercase) + '\n')

    naa = []
    for i in range(len(fastas)):
        arquivo.write('{:10}'.format(organismos[i]) + ' ')
        res = contaraa(fastas[i])
        arquivo.write(' '.join('{:3}'.format(str(int)) for int in res))
        arquivo.write('\n')
        naa.append(res)

    arquivo.write('\nb.Distância manhattan entre dois organismos\n')
    arquivo.write(comparar(naa, 'dmanhattan'))

    arquivo.write('\n\nc.Distância euclidiana entre dois organismos\n')
    arquivo.write(comparar(naa, 'deuclidiana'))

    arquivo.write('\n\nd.Distância supremum entre dois organismos\n')
    arquivo.write(comparar(naa, 'dsupremum'))

    arquivo.write('\n\nd.Similaridade de cosseno entre dois organismos\n')
    arquivo.write(comparar(naa, 'simcos'))
```
