# Tarefa 3 - Classificação de sementes de trigo

Este notebook tem como objetivo realizar a classificação de sementes de trigo utilizando diferentes algoritmos de aprendizado de máquina. Os métodos que serão aplicados incluem:

1. Árvore de Decisão
2. Naive Bayes
3. Support Vector Machine (SVM)

Vamos trabalhar com os dados de sementes de trigo, que contêm informações sobre as propriedades físicas das sementes e a classificação em três tipos diferentes.

O processo será dividido nas seguintes etapas:
1. Carregamento e pré-processamento dos dados
2. Divisão dos dados em treino e teste
3. Treinamento e teste com três algoritmos de classificação
4. Avaliação da acurácia dos modelos

```python
#Código de importação de bibliotecas
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, export_graphviz
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
```

## 1. Carregamento dos dados

Agora, vamos carregar os dados de sementes de trigo a partir de um arquivo CSV. O arquivo contém várias propriedades das sementes, como área, perímetro, compacidade, entre outras.

O primeiro passo será ler os dados e dar uma olhada nas primeiras instâncias.

```python
#1.Carregando os dados
arquivo='https://www.openml.org/data/get_csv/1592291/phpPrh7lv'
df = pd.read_csv(arquivo, names=['área','perímetro','compacidade','comprimento','largura','coeficiente de assimetria','comprimento do sulco da semente','classe'], sep=",")
df=df.drop(df.index[[0]])
```

## 2. Pré-processamento dos dados

Verificaremos se há valores ausentes nos dados e removeremos quaisquer instâncias que contenham valores faltantes. Esse passo é importante para garantir que os modelos de aprendizado de máquina possam ser treinados corretamente.


```python
#Remove linhas com alguma coluna vazia
df.dropna(subset = ['área','perímetro','compacidade','comprimento','largura','coeficiente de assimetria','comprimento do sulco da semente','classe'], inplace=True)

#Retorna linhas com alguma coluna vazia
is_NaN = df.isnull()
row_has_NaN = is_NaN.any(axis=1)
rows_with_NaN = df[row_has_NaN]
len(rows_with_NaN)
```

## 3. Separação dos dados em treino e teste

Separe 200 instâncias aleatoriamente como os dados de treino e 10 como dados de teste (train_test_split do módulo sklearn.model_selection).

--> Dividiremos os dados em duas partes: 200 instâncias serão usadas para treino e 10 instâncias serão usadas para teste. A divisão será feita de forma aleatória, garantindo que os dados de treino e teste sejam representativos do conjunto de dados total.


```python
x_train, x_test, y_train, y_test = train_test_split(df.drop('classe',axis=1), df['classe'], train_size=200, test_size=10, random_state=1)
```

## 4. Classificação com Árvore de Decisão

Neste passo, aplicaremos um modelo de árvore de decisão para classificar as sementes. A árvore de decisão é um algoritmo simples e interpretável que divide os dados em "ramos" com base em características dos dados.


```python
# Inicialização do algorito de classificação
clf = DecisionTreeClassifier()

# Treinando o modelo de arvore de decisão:
clf = clf.fit(x_train,y_train)

# Vamos testar nossa nova instância
test_list=x_test.values.tolist()
C = clf.predict(test_list)
P = clf.predict_proba(test_list)
```

### Avaliação da Árvore de Decisão

Vamos calcular a acurácia do modelo de árvore de decisão comparando as predições com as classes reais dos dados de teste. A acurácia é a proporção de predições corretas.


```python
# Mostrando o resultado da classificação
texto=texto+"\n\nCLASSIFICAÇÃO POR ÁRVORES DE DECISÃO\n"
for i in range(len(x_test)):
  texto=texto+"A instância de teste: {} foi classificada como a classe '{}' com probabilidade {}\n".format(test_list[i], C[i], P[i])

acuracia=0
for i,j in zip(C,y_test):
  if i==j:
    acuracia=acuracia+1
acuracia=acuracia/len(C)
texto=texto+'\nA acurácia é de {0:.2%}\n'.format(acuracia)

f"Acerto da Árvore de Decisão: {acuracia:.2%}"
```

## 5. Classificação com Naive Bayes

Agora, vamos aplicar o modelo de Naive Bayes. O Naive Bayes é um modelo probabilístico que utiliza a teoria de Bayes para fazer classificações baseadas nas probabilidades de cada classe.


```python
# Initialização do classificador baseado no Naive Bayes
gnb = GaussianNB()

# Ensinando o classificador usando conjunto de treino
gnb.fit(x_train,y_train)

# Vamos testar nossa nova instância usando k-vizinhos
C = gnb.predict(x_test)
P = gnb.predict_proba(x_test)
```

### Avaliação do Naive Bayes

Assim como fizemos para a árvore de decisão, vamos calcular a acurácia do modelo Naive Bayes comparando as predições com as classes reais.


```python
# Mostrando o resultado da classificação
texto=texto+"\n\nCLASSIFICAÇÃO POR NAIVE BAYES\n"
for i in range(len(x_test)):
  texto=texto+"A instância de teste: {} foi classificada como a classe '{}' com probabilidade {}\n".format(test_list[i], C[i], P[i])

acuracia=0
for i,j in zip(C,y_test):
  if i==j:
    acuracia=acuracia+1
acuracia=acuracia/len(C)
texto=texto+'\nA acurácia é de {0:.2%}\n'.format(acuracia)
f"Acerto do Naive Bayes: {acuracia:.2%}"
```

## 6. Classificação com SVM

Finalmente, aplicaremos um modelo de Support Vector Machine (SVM). O SVM é um algoritmo poderoso para classificação que tenta encontrar a melhor fronteira (hiperplano) entre as classes.


```python
# Initialização do classificador baseado no SVM
sup = SVC()

# Ensinando o classificador usando conjunto de treino
sup.fit(x_train,y_train)

# Vamos testar nossa nova instância usando k-vizinhos
C = sup.predict(x_test)
#P = sup.predict_proba(x_test)
```

### Avaliação do SVM

Vamos agora calcular a acurácia do modelo SVM. Assim como os modelos anteriores, a acurácia será medida pela comparação das predições com as classes reais dos dados de teste.


```python
# Mostrando o resultado da classificação
texto=texto+"\n\nCLASSIFICAÇÃO POR SVM\n"
for i in range(len(x_test)):
  texto=texto+"A instância de teste: {} foi classificada como a classe '{}'\n".format(test_list[i], C[i])

acuracia=0
for i,j in zip(C,y_test):
  if i==j:
    acuracia=acuracia+1
acuracia=acuracia/len(C)
texto=texto+'\nA acurácia é de {0:.2%}\n'.format(acuracia)
f"Acerto do SVM: {acuracia:.2%}"
```
