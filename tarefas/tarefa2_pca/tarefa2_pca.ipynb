## 1) Carregamento de Dados
Para carregar os dados usando o pacote sklearn você pode usar as duas linhas abaixo. O módulo datasets do pacote sklearn importa os conjuntos de dados para praticar as técnicas de mineração de dados. A função load_breast_cancer() além de carregar os dados fornece também os rótulos de classe.

--> Vamos carregar o conjunto de dados usando o método `load_breast_cancer` da biblioteca `sklearn.datasets`.

```python
from sklearn.datasets import load_breast_cancer
breast = load_breast_cancer()
```

## 2) Exploração de Dados
Os dados são disponibilizados no campo ".data" da varável "breast", enquanto os rótulos são armazenados no campo ".target". Verifique se os dados possuem exatamente 569 amostras com trinta características (atributos) e cada amostra tem um rótulo associado a ela. Coloque os dados na estrutura DataFrame do pacote pandas e mostra o começo e fim dos dados com a descrição de características.

--> Os dados possuem 569 amostras e 30 atributos. Vamos organizá-los em um DataFrame e inspecionar os primeiros e últimos registros.

```python
import pandas as pd
df = pd.DataFrame(breast.data, columns=breast.feature_names)
df['label'] = pd.Series(breast.target)
df['label'].replace(0, 'Benign', inplace=True)
df['label'].replace(1, 'Malignant', inplace=True)

print(df.head())
print(df.tail())

texto=texto+str(len(df))+' amostras'+'\n'
```

## 3) Pré-processamento - Padronização
Padronize todos os atributos usando a função StandardScaler() do módulo sklearn.preprocessing. Verifique se os atributos normalizados tem as médias próximas à zero e desvio padrão igual 1.

--> Vamos padronizar os dados usando `StandardScaler`, de forma que cada atributo tenha média 0 e desvio padrão 1.

```python
from sklearn import preprocessing
scaler = preprocessing.StandardScaler()
scaled = scaler.fit_transform(breast.data)
Xnorm = pd.DataFrame(scaled, columns=breast.feature_names)
texto=texto+'Os atributos normalizados tem as médias próximas a 0:'+str(all(round(Xnorm.mean(axis=0)))==0)+'\n'
texto=texto+'Os atributos normalizados tem desvio padrão igual a 1:'+str(all(round(Xnorm.std()))==1)+'\n'
```

## 4) PCA com 2 Componentes


Realize a analise de componentes principais usando objeto PCA do módulo sklearn.decomposition. Aqui Xnorm é a matriz de dados com atributos normalizados (30 colunas) e Xpca é a matriz de dados processados somente na direção das componentes principais (2 colunas). Use duas componentes como parâmetro e determine a quantidade de variação por componente usando o resultado armazenado em estrutura pca.explained_variance_ratio_. Qual porcentagem de variação essas duas componentes representam? Qual é a redução de quantidade de dados?

--> Vamos aplicar PCA com 2 componentes e verificar quanta variação cada uma representa. Também calcularemos a redução dimensional.

```python
from sklearn.decomposition import PCA
pca1 = PCA(n_components=2)
Xpca = pca1.fit_transform(Xnorm)
texto=texto+'2 componentes:'+str(pca1.explained_variance_ratio_)+'\n'
texto=texto+'Porcentagem de variação: {0:.2%} {1:.2%}'.format(*pca1.explained_variance_ratio_.tolist())+'\n'
texto=texto+'A redução de quantidade de dados: {0:.2%}'.format(1-(pca1.n_components/len(Xnorm.columns)))+'\n'
```

## 5) Visualização das Componentes Principais
Produza um gráfico de todas as instâncias dadas no espaço de duas coordenadas correspondentes as componentes principais, marcando as instâncias de dois rótulos de classes com cores diferentes. 

--> Vamos plotar os dados projetados nas duas componentes principais, colorindo por classe (Benign x Malignant).

```python
import matplotlib.pyplot as plt
targets = set(df['label'])
colors  = ['r', 'b']

for target, color in zip(targets,colors):
  indices = df['label'] == target
  plt.scatter(Xpca[indices,0], Xpca[indices,1], color=color, label=target)

plt.xlabel('componente principal 1')
plt.ylabel('componente principal 2')
plt.legend()

plt.savefig('grafico-pca.png')
plt.clf()
```

## 6) PCA com 3 Componentes
Repita a analise usando 3 componentes principais. Qual é o ganho de variação em relação a duas componentes?

--> Vamos repetir a análise com 3 componentes e observar se há ganho significativo na variância explicada.

```python
pca3 = PCA(n_components=3)
Xpca3 = pca3.fit_transform(Xnorm)

print("Variação explicada pelas 3 componentes:")
print(pca3.explained_variance_ratio_)
print("\nPorcentagem de variação:")
print("{0:.2%} {1:.2%} {2:.2%}".format(*pca3.explained_variance_ratio_))

ganho = (pca3.explained_variance_ratio_.sum() - pca1.explained_variance_ratio_.sum())
print(f"\nGanho de variação em relação a 2 componentes: {ganho:.2%}")
```
